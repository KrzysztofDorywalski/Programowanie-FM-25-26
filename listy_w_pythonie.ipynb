# 📘 Listy w Pythonie

Listy to **podstawowy typ danych** w Pythonie służący do przechowywania wielu elementów w jednej strukturze.

Można w nich przechowywać liczby, napisy, a nawet inne listy. Są **modyfikowalne (mutable)** i zachowują kolejność elementów.

## 1️⃣ Tworzenie list
Listę tworzymy używając nawiasów kwadratowych `[]` lub funkcji `list()`.

```python
# Tworzenie list
liczby = [1, 2, 3, 4, 5]
owoce = ["jabłko", "banan", "czereśnia"]
mieszana = [1, "tekst", True, 3.14]

print(liczby)
print(owoce)
print(mieszana)
```

## 2️⃣ Dostęp do elementów listy
- Indeksy zaczynają się od `0`.
- Możemy też używać indeksów ujemnych (od końca listy).

```python
owoce = ["jabłko", "banan", "czereśnia"]
print(owoce[0])   # pierwszy element
print(owoce[-1])  # ostatni element
```

## 3️⃣ Modyfikowanie listy
Możemy zmieniać elementy listy, dodawać i usuwać pozycje.

```python
owoce = ["jabłko", "banan", "czereśnia"]
owoce[1] = "gruszka"  # zmiana elementu
owoce.append("śliwka")  # dodanie na koniec
owoce.remove("jabłko")  # usunięcie elementu
print(owoce)
```

## 4️⃣ Najważniejsze metody list
- `append(x)` – dodaje element na koniec
- `insert(i, x)` – wstawia element na wskazanej pozycji
- `remove(x)` – usuwa pierwsze wystąpienie elementu
- `pop(i)` – usuwa element o danym indeksie i go zwraca
- `sort()` – sortuje listę
- `reverse()` – odwraca kolejność elementów
- `len(lista)` – zwraca długość listy

```python
liczby = [4, 1, 7, 3]
liczby.append(9)
liczby.sort()
print("Posortowane:", liczby)
liczby.reverse()
print("Odwrócone:", liczby)
print("Długość listy:", len(liczby))
```

## 5️⃣ Iteracja po liście
Możemy przechodzić po elementach listy pętlą `for`.

```python
owoce = ["jabłko", "gruszka", "śliwka"]
for owoc in owoce:
    print(f"Owoc: {owoc}")
```

## 6️⃣ Listy zagnieżdżone
Listy mogą zawierać inne listy – to tzw. **listy zagnieżdżone**.

```python
macierz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(macierz[0])      # pierwszy wiersz
print(macierz[1][2])   # element z drugiego wiersza, trzeciej kolumny
```
