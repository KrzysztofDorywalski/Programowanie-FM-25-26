# Tutorial: Praca z plikami w Pythonie

W tym notatniku poznasz podstawy pracy z plikami:
- otwieranie plików
- zapis do pliku
- odczyt z pliku
- korzystanie z konstrukcji `with`
- podstawy obsługi błędów podczas operacji na plikach


```python
# Tworzenie i zapis do pliku

# Otwieramy plik w trybie 'w' (write) – jeśli plik nie istnieje, zostanie utworzony
with open('przyklad.txt', 'w', encoding='utf-8') as f:
    f.write('To jest przykład pierwszej linijki.\n')
    f.write('A to jest druga linijka.')

print('Zapisano dane do pliku przyklad.txt')
```

## Odczyt danych z pliku
Możemy odczytać cały plik lub linia po linii.

```python
# Odczyt całej zawartości pliku

with open('przyklad.txt', 'r', encoding='utf-8') as f:
    content = f.read()

print('Zawartość pliku:')
print(content)
```

## Odczyt linia po linni

```python
# Czytanie pliku linia po linii

with open('przyklad.txt', 'r', encoding='utf-8') as f:
    for line in f:
        print('Linia:', line.strip())
```

## Obsługa błędów przy pracy z plikami

```python
# Obsługa błędów przy próbie otwarcia nieistniejącego pliku

try:
    with open('nieistnieje.txt', 'r', encoding='utf-8') as f:
        data = f.read()
except FileNotFoundError:
    print('Plik nie istnieje! Upewnij się, że ścieżka jest poprawna.')
```
